```python
posts_df['BENEFIT_SUBCATEGORY_NAME'].value_counts()
```

```python
# get a list of unique benefit subcategories
benefit_subcategories = posts_df['BENEFIT_SUBCATEGORY_NAME'].unique()
print(benefit_subcategories)
```

```python
benefit_subcategories = [str(x).strip() for x in benefit_subcategories if pd.notnull(x)]
# remove \n
benefit_subcategories = [x.replace('\n', '') for x in benefit_subcategories]
print(benefit_subcategories)
benefit_subcategories = [x for x in benefit_subcategories if x != 'nan' and x != 'None' and x != '']
print(benefit_subcategories)
# benefit_subcategories = set(benefit_subcategories)
# print(benefit_subcategories)
```
