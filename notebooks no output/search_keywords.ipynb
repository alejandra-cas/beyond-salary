```python
import pandas as pd 
```

```python
even_sample = pd.read_parquet('data/salary_sample_body.parquet.gzip')
```

```python
# search for "bonding leave" in BODY column of even_sample
bonding_leave_results = even_sample[even_sample['BODY'].str.contains("bonding leave", case=False, na=False)]
# print only the phrases that contain "bonding leave"
bonding_leave_phrases = bonding_leave_results['BODY'].str.extract(r'([^.!?]*bonding leave[^.!?]*)', expand=False)
print(bonding_leave_phrases.dropna().values)
```

```python
# search for "bonding leave" in BODY column of even_sample
diversity_results = even_sample[even_sample['BODY'].str.contains("diversity", case=False, na=False)]
# print only the phrases that contain "bonding leave"
diversity_phrases = diversity_results['BODY'].str.extract(r'([^.!?]*diversity[^.!?]*)', expand=False)
for phrase in diversity_phrases.dropna().values:
    print(phrase)
    print('---')
```
