```python
import pandas as pd
```

```python
data = pd.read_parquet('../data/us_10m_nointernship_ai_skills_benefits.parquet.gzip')
```

```python
pd.set_option('display.max_columns', None)
```

```python
dates = pd.read_parquet('../data/US_10M_POSTS_DATES.parquet.gzip')
```

```python
data = data.merge(dates, on='ID', how = 'left')
```

```python
data[data['EXPIRED'].isna()]
```

```python
len(data)
```

```python
data[data['DURATION'].isna()]
```

```python
# get EXPIRED data type
data['EXPIRED'] = pd.to_datetime(data['EXPIRED'])
```

```python
data['POSTED'] = pd.to_datetime(data['POSTED'])
```

```python
data['DURATION_CALC'] = data['EXPIRED'] - data['POSTED']
```

```python
data['DURATION_CALC']
```

```python
data['DURATION_CALC'].describe()
```

```python
data['DURATION_CALC'] = pd.to_timedelta(data['DURATION_CALC']).dt.total_seconds() / (24 * 3600)
```

```python
# histogram with title and x labels
import matplotlib.pyplot as plt
plt.title('Duration of job postings')
plt.xlabel('Duration in days')
plt.ylabel('Number of job postings')
data['DURATION_CALC'].hist(grid = False)
```

```python
data.to_parquet('../data/us_10m_nointernship_ai_skills_benefits.parquet.gzip', compression='gzip')
```

```python
data[['POSTED','EXPIRED','DURATION', 'DURATION_CALC']]
```

```python
data.groupby('YEAR')['DURATION_CALC'].mean()
```

```python
# plot the average duration of job postings per year per AI role
data.groupby(['YEAR', 'AI ROLE'])['DURATION_CALC'].mean().unstack().plot()
```
